## Gen AI: Autocomplete Several Tasks

```python
# Write a class that contains several functions for loops, conditional statements, data structures, and more respectively

class MyClass:
    def loop_example(self):
        for i in range(5):
            print(f"Loop iteration: {i}")

    def conditional_example(self, number):
        if number > 0:
            print(f"{number} is positive.")
        elif number < 0:
            print(f"{number} is negative.")
        else:
            print("The number is zero.")

    def data_structure_example(self):
        my_list = [1, 2, 3, 4, 5]
        my_dict = {'a': 1, 'b': 2, 'c': 3}
        print("List:", my_list)
        print("Dictionary:", my_dict)

# Example usage of the class
my_instance = MyClass()
my_instance.loop_example()
my_instance.conditional_example(10)
my_instance.conditional_example(-5)
my_instance.conditional_example(0)
my_instance.data_structure_example()
```
