# Sets and Deques

```python
# Creating a set
my_set = {1, 2, 3, 4, 4, 5}
print(my_set)  # Output: {1, 2, 3, 4, 5} - duplicates are removed
```

```python
my_set.add(6)  # Adding an element
print(my_set)  # Output: {1, 2, 3, 4, 5, 6}
```

```python
my_set.remove(3)  # Removing an element
print(my_set)  # Output: {1, 2, 4, 5
```

```python
# Membership testing 
print(2 in my_set)
```

```python
print(7 in my_set)  # Output: False - 7 is not in the set
```

#### Deques

```python
from collections import deque
```

```python
# Creating a deque
my_deque = deque([1, 2, 3])
```

```python
# Adding elements to the right
my_deque.append(4)
print(my_deque)  # Output: deque([1, 2, 3, 4])
```

```python
# Adding elements to the left
my_deque.appendleft(0)
print(my_deque)  # Output: deque([0, 1, 2, 3, 4])
```

```python
# Removing elements from the right
my_deque.pop()
print(my_deque)  # Output: deque([0, 1, 2, 3])
```

```python
# Removing elements from the left
my_deque.popleft()
print(my_deque)  # Output: deque([1, 2, 3])
```

#### Real-world Scenario

```python
# Simplified example
undo_stack = deque()
redo_stack = deque()
```

```python
def perform_action(action):
    undo_stack.append(action)
    redo_stack.clear()  # Clear redo stack on new action    

def undo():
    if undo_stack: # Check if there are actions to undo
        action = undo_stack.pop()  # Pop the last action from the undo stack
        redo_stack.append(action) # 
        print(f"Undo: {action}")
    else:
        print("No actions to undo.")

def redo():
    if redo_stack: # Check if there are actions to redo
        action = redo_stack.pop() 
        undo_stack.append(action) # add the action back to the undo stack
        print(f"Redo: {action}")
    else:
        print("No actions to redo.")
```

```python
split_words = str("Hello world").split()
for x in split_words:
    perform_action(x)
```
